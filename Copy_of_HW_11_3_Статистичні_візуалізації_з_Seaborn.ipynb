## Завдання 3: Box Plot порівняння - Pandas vs Seaborn

**Завдання:**
Побудуйте box plot для кількості погодинних оренд велосипедів за погодними умовами з Pandas та Seaborn.

**Функція Seaborn: `sns.boxplot()`**

Можна побудувати окремо два графіки. Але для тих, хто хоче складніше - побудуйте ці 2 графіки на 1 фігурі.

Просунуте доповнення:
- підпишіть погодні умови їх інтерпретацією з опису даних в обох графіках

**Дайте відповідь на питання:**
- Яка візуальна різниця між Pandas та Seaborn бокс-плотами?

```python
import seaborn as sns


weather_names = {
    1: 'Ясно',
    2: 'Туман / хмарно',
    3: 'Легкий дощ / сніг',
    4: 'Сильний дощ / сніг'
}

df['weather_name'] = df['weather'].map(weather_names)


fig, axes = plt.subplots(1, 2, figsize=(16, 6))


df.boxplot(
    column='count',
    by='weather_name',
    ax=axes[0]
)

axes[0].set_title('Кількість оренд за погодними умовами — Pandas')
axes[0].set_xlabel('Погодні умови')
axes[0].set_ylabel('Кількість оренд')


fig.suptitle('')


sns.boxplot(
    data=df,
    x='weather_name',
    y='count',
    ax=axes[1]
)

axes[1].set_title('Кількість оренд за погодними умовами — Seaborn')
axes[1].set_xlabel('Погодні умови')
axes[1].set_ylabel('Кількість оренд')

plt.tight_layout()
plt.show()
```

В цілому обидві візуалізації схожі між собою, але друга більш сучасна і, на мою думку, якось більш зрозуміла, хоча набір даних однаковий.

## Завдання 5: Violin Plot для глибокого аналізу розподілів

**Завдання:**
Створіть violin plot для аналізу розподілу оренди за кварталами.

Дайте відповіді на питання:

**Питання для інтерпретації:**
1. Що показує "товщина" violin plot?
2. В якому кварталі найбільша варіабельність оренди?

1. Що показує "товщина" violin plot?
Вона показує, що уцьому діапазоні зосереджена найбіща кількість даних. Чим ширша частина - тим більше значень у ньому.
2. В якому кварталі найбільша варіабельність оренди?
У 3 кварталі, адже розподіл має великий розмах.
3. Яка перевага violin plot над звичайним box plot?
Violin plot показує не тільки медіану, квартилі та викиди, а й форму розподілу даних та їхню щільність.

## Завдання 7: Joint Plot для детального аналізу двох змінних

**Завдання:**
Проаналізуйте залежність між температурою та орендою за допомогою joint plot. В якості візуальної розбивки за категоріями (параметр `hue`) додайте `workingday`.

Дайте відповіді на питання:

**Питання для інтерпретації:**
1. Що показують графіки по краях?
2. Чи є різниця у поведінці користувачів у робочий і неробочий день?

```python
g = sns.jointplot(
    data=df,
    x='temp',
    y='count',
    hue='workingday',
    height=8,
    alpha=0.4
)

g.fig.suptitle(
    'Залежність кількості оренд від температури',
    y=1.02
)

plt.show()
```

1. Що показують графіки по краях?
Графіки по краях показують розподіл кожної окремої змінної. Верхній графік показує розподіл температури, а правий — розподіл кількості оренд.

2. Чи є різниця у поведінці користувачів у робочий і неробочий день?
Так, різниця помітна. У робочі дні спостерігається більше значень із високою кількістю оренд, особливо при середній та високій температурі. У неробочі дні значення оренд частіше зосереджені на нижчому рівні. Це видно як із розподілу точок, так і з графіка розподілу count праворуч.
