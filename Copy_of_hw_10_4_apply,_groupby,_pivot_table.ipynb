# **Що найчастіше купують?**

```python
df.groupby(['Customer type', 'Product line'])['Quantity'].sum().unstack()
```

```python
df.groupby(['Customer type', 'Product line'])['Quantity'].sum().groupby(level=0).idxmax()
```

За цими даними можу запропонувати спеціальні діскаунти для мемберів в категоріях, в яких більш за все покупок, а для нормальних - систему залучення до програми лояльності, які дають чкісь корисні бенефіти.

# **Коли купують найбільше?**

```python
df['Date'] = pd.to_datetime(df['Date'])
df['Day'] = df['Date'].dt.day_name()
```

```python
df.groupby('Day')['Quantity'].mean().sort_values(ascending=False).round(2)
```

Запускати спеціальні акції або пропозиції у дні, коли найменше продажів за нашими даними, тобто у поеділок та п'ятницю, а також оптимізувати працю у дні з найбільшою загруженістю.

# **Наскільки клієнти задоволені тим, що вони купують?**

```python
avg_rating_by_branch = df.groupby('Branch')['Rating'].mean().sort_values(ascending=False).round(2)
print(avg_rating_by_branch)
```

```python
avg_rating_by_product = df.groupby('Product line')['Rating'].mean().sort_values(ascending=False).round(2)
print(avg_rating_by_product)
```

```python
avg_rating_combo = df.groupby(['Branch', 'Product line'])['Rating'].mean().unstack().round(2)
print(avg_rating_combo)
```

За цими ланими можна побасити степінь задоволеності клієнтами конкретних товарів. Можливо перегляд асортименту або постачальників.
Якщо низький рейтинг філії - варто переглянути рівень сервісу (швидкість, кваліфікованість або чистота).
